```python
# 1. Imports
import pandas as pd
import numpy as np
from pathlib import Path

# 2. Paths
RAW = Path("../data/raw")
PROCESSED = Path("../data/processed")

# 3. Load data (use pyarrow or low_memory=False)
train_transaction = pd.read_csv(RAW / "train_transaction.csv")
train_identity = pd.read_csv(RAW / "train_identity.csv")

print(train_transaction.shape)
print(train_identity.shape)

# 4. Merge
train = train_transaction.merge(train_identity, how="left", on="TransactionID")

# 5. Basic info
print(train.info())
print(train["isFraud"].value_counts(normalize=True))

# 6. Save merged version (parquet is much faster)
train.to_parquet(PROCESSED / "train_merged.parquet", index=False)
```
